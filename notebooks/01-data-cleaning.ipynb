```python
from pathlib import Path


DATA_PATH = Path(r"..\data\raw form\used_cars_data.csv")


```

```python
import pandas as pd

df = pd.read_csv(DATA_PATH)

print(df.shape)
df.head()
```

```python
print("Dataset shape:", df.shape)
print("\nColumn names:", df.columns.tolist())
print("\nFirst five records:")
display(df.head())
```

```python
df.info
```

```python
missing_summary = pd.DataFrame({ 
    "missing_count": df.isna().sum(), 
    "missing percentage": (df.isna().sum() / len(df) * 100).round(2)
    })
missing_summary = missing_summary.sort_values(by="missing_count", ascending=False)
missing_summary
```

```python
df.columns = (df.columns.str.strip().str.lower().str.replace(" ", "_").str.replace(".","", regex=False))
print(df.columns.tolist())
```

```python
df = df.drop(columns=["sno"], errors="ignore")
print(df.columns.tolist())
```

```python
import numpy as np

df["brand"] = df["name"].str.split().str[0]
df["mileage_kmpl"] = (
    df["mileage"]
    .astype("string")
    .str.extract(r"([\d.]+)", expand=False)
    .astype(float)
)
df["engine_cc"] = (
    df["engine"]
    .astype("string")
    .str.extract(r"([\d.]+)", expand=False)
    .astype(float)
)
df["power_bhp"] = (
    df["power"]
    .astype("string")
    .str.extract(r"([\d.]+)", expand=False)
    .astype(float)
)
df["price"] = pd.to_numeric(df["price"], errors="coerce")

REFERENCE_YEAR = 2020
df["vehicle_age"] = REFERENCE_YEAR - df["year"]
display(
    df[
        [
            "name",
            "brand",
            "year",
            "vehicle_age",
            "mileage",
            "mileage_kmpl",
            "engine",
            "engine_cc",
            "power",
            "power_bhp",
            "seats",
            "price",
        ]
    ].head(10)
)
```

```python
df_train = df.dropna(subset=["price"]).copy()
print("Raw dataset rows:", len(df))
print("Rows with known prices:", len(df_train))
print("Rows excluded because price is missing:", len(df) - len(df_train))
```

```python
df_unpriced = df[df["price"].isna()].copy()
print("Unpriced records:", len(df_unpriced))
```

```python
zero_mileage = df_train[df_train["mileage_kmpl"] <= 0]
print("Zero or negative mileage records:", len(zero_mileage))

display(
    zero_mileage[
        [
            "name",
            "year",
            "fuel_type",
            "mileage",
            "mileage_kmpl",
            "engine_cc",
            "power_bhp",
            "price",
        ]
    ].head(20)
)
```

```python
df_train.loc[df_train["mileage_kmpl"] <= 0, "mileage_kmpl"] = np.nan
```

```python
numeric_columns = [
    "year",
    "kilometers_driven",
    "mileage_kmpl",
    "engine_cc",
    "power_bhp",
    "seats",
    "vehicle_age",
    "price",
]

df_train[numeric_columns].describe().T
```

```python
quality_checks = pd.Series({
    "Training rows": len(df_train),
    "Rows missing mileage_kmpl": df_train["mileage_kmpl"].isna().sum(),
    "Rows missing engine_cc": df_train["engine_cc"].isna().sum(),
    "Rows missing power_bhp": df_train["power_bhp"].isna().sum(),
    "Rows missing seats": df_train["seats"].isna().sum(),
    "Zero or negative mileage": (df_train["mileage_kmpl"] <= 0).sum(),
    "Zero or negative price": (df_train["price"] <= 0).sum(),
    "Negative vehicle age": (df_train["vehicle_age"] < 0).sum(),
    "Exact duplicate rows": df_train.duplicated().sum(),
}).sort_values(ascending=False)

quality_checks
```

```python
import numpy as np

# 1. Treat zero or negative mileage as invalid/missing.
df_train.loc[df_train["mileage_kmpl"] <= 0, "mileage_kmpl"] = np.nan

# 2. A listed vehicle cannot have zero or negative seats.
df_train.loc[df_train["seats"] <= 0, "seats"] = np.nan

# 3. Engine displacement below 500 cc is not credible for this dataset.
# Inspect the affected records before keeping this threshold permanently.
df_train.loc[df_train["engine_cc"] < 500, "engine_cc"] = np.nan

# 4. Kilometres driven above 500,000 are treated as likely data-entry errors.
# Preserve the record but remove the implausible odometer value.
df_train.loc[
    df_train["kilometers_driven"] > 500_000,
    "kilometers_driven"
] = np.nan
```

```python
display(
    df_train.loc[
        df_train["kilometers_driven"].isna(),
        [
            "name",
            "year",
            "kilometers_driven",
            "fuel_type",
            "transmission",
            "price"
        ]
    ]
)
```

```python
df_train["high_kilometers_flag"] = (
    df_train["kilometers_driven"] > 500_000
)

display(
    df_train.loc[
        df_train["high_kilometers_flag"],
        [
            "name",
            "year",
            "kilometers_driven",
            "fuel_type",
            "transmission",
            "price"
        ]
    ]
)
```

```python
df_train.loc[
    df_train["high_kilometers_flag"],
    "kilometers_driven"
] = np.nan
```

```python
display(
    df_train.loc[
        df_train["engine_cc"].notna() & (df_train["engine_cc"] < 500),
        [
            "name",
            "year",
            "engine",
            "engine_cc",
            "fuel_type",
            "price"
        ]
    ]
)
```

```python
df_train["low_engine_flag"] = (
    df_train["engine_cc"].notna()
    & (df_train["engine_cc"] < 500)
)

df_train.loc[df_train["low_engine_flag"], "engine_cc"] = np.nan
```

```python
display(
    df_train.loc[
        df_train["seats"].notna() & (df_train["seats"] <= 0),
        [
            "name",
            "year",
            "seats",
            "price"
        ]
    ]
)
```

```python
df_train["invalid_seats_flag"] = (
    df_train["seats"].notna()
    & (df_train["seats"] <= 0)
)

df_train.loc[df_train["invalid_seats_flag"], "seats"] = np.nan
```

```python
model_columns = [
    "name",
    "brand",
    "location",
    "year",
    "vehicle_age",
    "kilometers_driven",
    "fuel_type",
    "transmission",
    "owner_type",
    "mileage_kmpl",
    "engine_cc",
    "power_bhp",
    "seats",
    "price",
]

df_model = df_train[model_columns].copy()

print("Final modelling dataset shape:", df_model.shape)

display(df_model.head())

print("\nRemaining missing values:")
display(
    df_model.isna()
    .sum()
    .sort_values(ascending=False)
)
```

## Data-cleaning decisions

- Removed the serial-number field because it is an identifier, not a vehicle attribute.
- Excluded `new_price` from modelling because 86.13% of values were missing.
- Excluded 1,234 observations with missing `price` only from supervised model training because price is the prediction target.
- Extracted numeric values from mileage, engine, and power fields.
- Converted 0.0 kmpl mileage values to missing because they are implausible and appear to encode unavailable data.
- Replaced invalid seat counts and implausible engine/odometer values with missing values rather than deleting the full listing.
- Deferred numeric missing-value imputation until model training, where it will be fitted only on training data.

```python
from pathlib import Path

OUTPUT_PATH = Path(r"..\data\processed form\used_cars_cleaned.csv")
OUTPUT_PATH.parent.mkdir(parents=True, exist_ok=True)

df_model.to_csv(OUTPUT_PATH, index=False)

print(f"Saved cleaned data to: {OUTPUT_PATH.resolve()}")
```
